## 1 · Variables, Datatypes & Operators

Python is **dynamically typed** — you don't declare types, but every value *has* one. The core built-in types:

| Type | Example | Use |
|------|---------|-----|
| `str` | `"TARS"` | text |
| `int` | `45` | whole numbers |
| `float` | `2.2` | decimals |
| `bool` | `True` / `False` | logic |
| `NoneType` | `None` | "no value" |

````python
name = "TARS"          # str
total_days = 45        # int
progress = 2.2         # float
is_shipping = True     # bool
notes = None           # NoneType

for value in (name, total_days, progress, is_shipping, notes):
    print(f"{str(value):<8} -> {type(value).__name__}")
````

````python
# Arithmetic operators
print(7 + 2, 7 - 2, 7 * 2)
print(7 / 2)    # true division  -> 3.5
print(7 // 2)   # floor division -> 3
print(7 % 2)    # modulo         -> 1
print(7 ** 2)   # power          -> 49

# Comparison + logical operators return bools
done = (total_days == 45) and is_shipping
print("done?", done)

# f-strings: the standard way to format text
print(f"{name} ships in {total_days} days ({progress}% complete)")
````

## 2 · Conditionals & Loops

Control flow decides *what* runs and *how often*. Python uses **indentation** (4 spaces) to define blocks — there are no braces.

````python
def phase_for(day):
    """Return the bootcamp phase name for a given day number."""
    if day <= 3:
        return "Foundations"
    elif day <= 7:
        return "How LLMs Work"
    elif day <= 11:
        return "Building LLM Apps"
    else:
        return "Advanced (RAG, Agents, Ops...)"

for day in [1, 5, 9, 20]:
    print(f"Day {day:>2}: {phase_for(day)}")
````

````python
# while loop, break and continue
day = 0
while True:
    day += 1
    if day % 2 == 0:
        continue          # skip even days
    if day > 7:
        break             # stop after day 7
    print("Odd study day:", day)
````

## 3 · Data Structures

Choosing the right container is half of clean Python.

| Structure | Ordered | Mutable | Duplicates | Syntax |
|-----------|---------|---------|------------|--------|
| **list**  | ✅ | ✅ | ✅ | `[1, 2, 3]` |
| **tuple** | ✅ | ❌ | ✅ | `(1, 2, 3)` |
| **set**   | ❌ | ✅ | ❌ | `{1, 2, 3}` |
| **dict**  | ✅* | ✅ | keys unique | `{"a": 1}` |

<sub>*dicts preserve insertion order since Python 3.7</sub>

````python
# LIST - ordered, mutable
topics = ["setup", "variables", "loops"]
topics.append("data structures")
topics[0] = "environments"
print("list  :", topics, "| len:", len(topics))
print("slice :", topics[1:3])      # items 1 and 2

# TUPLE - ordered, immutable (great for fixed records / coordinates)
point = (10, 20)
x, y = point                        # tuple unpacking
print("tuple :", point, "-> x =", x, "y =", y)
````

````python
# SET - unique, unordered (perfect for de-duplication & membership tests)
tools = ["vscode", "conda", "uv", "conda", "vscode"]
unique_tools = set(tools)
print("set        :", unique_tools)
print("is uv in?  :", "uv" in unique_tools)
print("union      :", {"uv", "pip"} | {"uv", "poetry"})
print("intersect  :", {"uv", "pip"} & {"uv", "poetry"})
````

````python
# DICT - key -> value lookups
student = {"name": "Ada", "scores": [82, 91, 77]}
print("name   :", student["name"])
print("get    :", student.get("email", "<missing>"))   # safe lookup
student["email"] = "ada@tars.ai"                        # add a key

for key, value in student.items():
    print(f"  {key:<7}: {value}")
````

## 4 · Comprehensions

A comprehension builds a new collection in one readable line: `[expr for item in iterable if condition]`.

````python
# list comprehension
even_squares = [n ** 2 for n in range(10) if n % 2 == 0]
print("even squares :", even_squares)

# dict comprehension
lengths = {word: len(word) for word in ["rag", "agent", "transformer"]}
print("lengths      :", lengths)

# set comprehension
first_letters = {word[0] for word in ["conda", "uv", "colab"]}
print("first letters:", first_letters)
````
